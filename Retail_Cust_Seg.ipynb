## Customer Segmentation (Machine Learning)

The EDA above showed *(How)* customers differ on individual variables. This section builds the actual segmentation model requested in the brief: a clustering model that groups customers into distinct segments based on combined purchasing behaviour and demographic characteristics, followed by cluster evaluation and profiling.

### Data Cleaning

```python
df.head()
```

```python
df.tail()
```

```python
df.isnull().sum()
```

```python
# I drop Income because it has a small number of missing values (~1% of rows)
df = df.dropna(subset=['Income'])
```

```python
# Remove unrealistic outliers (a few Year_Birth entries go back to the 1800s, and a couple of Income values are extreme outliers that would distort scaling)
df = df[df['Age'] < 100]
df = df[df['Income'] < 200000]
```

```python
print("After Cleaning the Dataset, the shape of the Dataset is now: ", df.shape)
```

### Feature Engineering

We combine the raw columns into features that better represent customer *behaviour*, which is what we want the model to segment on.

```python
# Number of dependents at home
df['Total_Children'] = df['Kidhome'] + df['Teenhome']
```

```python
# Total amount spent across all product categories
df['Total_Spending'] = df[products].sum(axis=1)
```

```python
# Total number of purchases across all channels (deals, web, catalog, store)
df['Total_Purchases'] = (df['NumDealsPurchases'] + df['NumWebPurchases']
                          + df['NumCatalogPurchases'] + df['NumStorePurchases'])
```

```python
# How long the customer has been enrolled, in days (relative to the most recent enrollment date)
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], dayfirst=True, format='mixed')
df['Customer_Tenure_Days'] = (df['Dt_Customer'].max() - df['Dt_Customer']).dt.days

df[['Total_Children', 'Total_Spending', 'Total_Purchases', 'Customer_Tenure_Days']].describe()
```

### Selecting Clustering Features & Scaling

We pick a compact set of features that together capture *who the customer is* (age, income, family size), *how they behave* (spending, purchases, tenure, site visits), and *how they respond* (campaigns accepted). Clustering algorithms are distance-based, so we standardize everything to the same scale first.

```python
cluster_features = [
    'Age', 'Income', 'Total_Children', 'Recency',
    'Total_Spending', 'Total_Purchases', 'NumWebVisitsMonth',
    'Customer_Tenure_Days', 'Total_Accepted_Cmp'
]

X = df[cluster_features].copy() # here we are trying to make use of the copy of the Dataset
X.describe()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=cluster_features)
X_scaled.head()
```

### Choosing the Number of Clusters

We use the elbow method (inertia) together with the silhouette score to pick a sensible number of segments. The elbow tells us where adding more clusters stops meaningfully reducing within-cluster variance; the silhouette score tells us how well-separated the clusters actually are.

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

inertia = []
silhouette_scores = []
k_range = range(2, 11)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertia.append(km.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, labels))

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
axes[0].plot(list(k_range), inertia, marker='o')
axes[0].set_title('Elbow Method')
axes[0].set_xlabel('Number of Clusters (k)')
axes[0].set_ylabel('Inertia')

axes[1].plot(list(k_range), silhouette_scores, marker='o', color='green')
axes[1].set_title('Silhouette Score by k')
axes[1].set_xlabel('Number of Clusters (k)')
axes[1].set_ylabel('Silhouette Score')
plt.tight_layout()
plt.show()

for k, s in zip(k_range, silhouette_scores):
    print(f"k={k}: silhouette={s:.4f}")
```

```python
optimal_k = 4 

kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)

print(df['Cluster'].value_counts().sort_index())
print(f"Silhouette Score: {silhouette_score(X_scaled, df['Cluster']):.4f}")
```

### Visualising the Segments (PCA)

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 7))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df['Cluster'], cmap='viridis', alpha=0.6)
plt.title('Customer Segments (PCA Projection)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.colorbar(scatter, label='Cluster')
plt.show()

print(f"Explained variance ratio: {pca.explained_variance_ratio_}")
print(f"Total variance explained by 2 components: {pca.explained_variance_ratio_.sum():.2%}")
```

### Cluster Profiling

```python
cluster_profile = df.groupby('Cluster')[cluster_features].mean().round(2)
cluster_profile['Count'] = df['Cluster'].value_counts().sort_index()
cluster_profile
```

```python
fig, axes = plt.subplots(3, 3, figsize=(18, 14))
axes = axes.flatten()
for i, feat in enumerate(cluster_features):
    df.groupby('Cluster')[feat].mean().plot(kind='bar', ax=axes[i], color='skyblue')
    axes[i].set_title(feat)
    axes[i].set_xlabel('Cluster')
plt.tight_layout()
plt.show()
```

**Interpreting the segments:**

- **Cluster 0 - "Established Steady Spenders" (610 customers, ~27%):** Oldest group (avg. age 59.8), moderate income (~$57.9k), solid spending (~$791.70) and purchase activity, but by far the longest customer tenure (455 days). Long-standing, consistent buyers who rarely respond to campaigns (0.20 acceptance).

- **Cluster 1 - "Low-Value Browsers" (1021 customers, ~45%, largest segment):** Lowest income (~$34.2k) and lowest spending by a wide margin (~$100.60), fewest purchases (8.0), yet the highest web visit frequency (6.50/month). They browse but rarely buy, and are the least responsive to campaigns (0.09 acceptance).

- **Cluster 2 - "High-Value Campaign Responders" (128 customers, ~6%, smallest segment):** Highest income (~$80.7k) and highest spending (~$1590.98), with a campaign acceptance rate (2.51) far above every other segment. This is the premium, marketing-responsive segment worth prioritizing for targeted offers.

- **Cluster 3 - "Affluent but Disengaged" (453 customers, ~20%):** High income (~$75.7k) and high spending (~$1222.91) similar to Cluster 2, but the lowest web engagement (2.28 visits/month) and low campaign response (0.28). They spend well but tune out marketing outreach -- worth investigating why (loyalty program members? in-store preference?).

### Model Evaluation Summary

```python
from sklearn.metrics import calinski_harabasz_score, davies_bouldin_score
```

```python
sil = silhouette_score(X_scaled, df['Cluster'])
ch = calinski_harabasz_score(X_scaled, df['Cluster'])
db = davies_bouldin_score(X_scaled, df['Cluster'])

print("Model Evaluation")
print("-" * 30)
print(f"Number of clusters (k): {optimal_k}")
print(f"Silhouette Score:        {sil:.4f}   (higher is better, range -1 to 1)")
print(f"Calinski-Harabasz Score: {ch:.2f}   (higher is better)")
print(f"Davies-Bouldin Score:    {db:.4f}   (lower is better)")
```
